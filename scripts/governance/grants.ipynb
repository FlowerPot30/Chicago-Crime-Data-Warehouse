# Grants privileges

```python
%sql
-- data_engineers: can access catalog and do everything in every layer

GRANT USAGE ON CATALOG chicago_crime TO `data_engineers`;
GRANT ALL PRIVILEGES ON SCHEMA chicago_crime.bronze TO `data_engineers`;
GRANT ALL PRIVILEGES ON SCHEMA chicago_crime.silver TO `data_engineers`;
GRANT ALL PRIVILEGES ON SCHEMA chicago_crime.gold TO `data_engineers`;

-- analysts: can only access and read on schema gold

GRANT USAGE ON CATALOG chicago_crime TO `analysts`;
GRANT USAGE ON SCHEMA chicago_crime.gold TO `analysts`;
GRANT SELECT ON SCHEMA chicago_crime.gold TO `analysts`
```
